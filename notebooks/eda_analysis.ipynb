### 3.1. Distance Analysis - Critical Missing Feature 🎯

<h3 style="color:#007bff;">Calculate Haversine Distance</h3>

<p>The individual coordinates showed weak correlation with delivery time. However, the <strong>actual travel distance</strong> between store and drop location is expected to be a strong predictor. Let's calculate it using the Haversine formula.</p>

```python
from math import radians, sin, cos, sqrt, atan2

def haversine_distance(lat1, lon1, lat2, lon2):
    """
    Calculate the great circle distance between two points 
    on the earth (specified in decimal degrees)
    Returns distance in kilometers
    """
    # Convert decimal degrees to radians
    lat1, lon1, lat2, lon2 = map(radians, [lat1, lon1, lat2, lon2])
    
    # Haversine formula
    dlat = lat2 - lat1
    dlon = lon2 - lon1
    a = sin(dlat/2)**2 + cos(lat1) * cos(lat2) * sin(dlon/2)**2
    c = 2 * atan2(sqrt(a), sqrt(1-a))
    
    # Radius of earth in kilometers
    R = 6371
    distance = R * c
    
    return distance

# Calculate distance for each delivery
df['Distance_km'] = df.apply(
    lambda row: haversine_distance(
        row['Store_Latitude'], row['Store_Longitude'],
        row['Drop_Latitude'], row['Drop_Longitude']
    ), axis=1
)

print("Distance Statistics:")
print(df['Distance_km'].describe())
print(f"\n✅ Distance feature created successfully!")
print(f"Range: {df['Distance_km'].min():.2f} km to {df['Distance_km'].max():.2f} km")
```

### 4.1. Statistical Hypothesis Testing 📊

<h3 style="color:#007bff;">ANOVA: Weather Impact on Delivery Time</h3>

```python
from scipy.stats import f_oneway

# Separate delivery times by weather
weather_groups = [df[df['Weather'] == w]['Delivery_Time'].values for w in df['Weather'].unique()]

# Perform ANOVA
f_stat, p_value = f_oneway(*weather_groups)

print("=" * 70)
print("HYPOTHESIS TEST 1: Does Weather Significantly Affect Delivery Time?")
print("=" * 70)
print(f"\n📝 Null Hypothesis (H0): All weather conditions have the same mean delivery time")
print(f"📝 Alternative Hypothesis (H1): At least one weather condition differs significantly\n")
print(f"🔬 Test: One-Way ANOVA")
print(f"📊 F-statistic: {f_stat:.4f}")
print(f"📊 P-value: {p_value:.10f}")
print(f"\n🎯 Decision: {'REJECT H0' if p_value < 0.05 else 'FAIL TO REJECT H0'} (α = 0.05)")
print(f"💡 Conclusion: Weather {'SIGNIFICANTLY' if p_value < 0.05 else 'DOES NOT significantly'} affects delivery time\n")

# Show mean delivery time by weather
weather_means = df.groupby('Weather')['Delivery_Time'].mean().sort_values()
print("Average Delivery Time by Weather:")
for weather, time in weather_means.items():
    print(f"  {weather:15s}: {time:.2f} minutes")
print("=" * 70)
```

<h3 style="color:#007bff;">ANOVA: Traffic Impact on Delivery Time</h3>

```python
# Separate delivery times by traffic
traffic_groups = [df[df['Traffic'] == t]['Delivery_Time'].values for t in df['Traffic'].unique()]

# Perform ANOVA
f_stat, p_value = f_oneway(*traffic_groups)

print("=" * 70)
print("HYPOTHESIS TEST 2: Does Traffic Significantly Affect Delivery Time?")
print("=" * 70)
print(f"\n📝 Null Hypothesis (H0): All traffic conditions have the same mean delivery time")
print(f"📝 Alternative Hypothesis (H1): At least one traffic condition differs significantly\n")
print(f"🔬 Test: One-Way ANOVA")
print(f"📊 F-statistic: {f_stat:.4f}")
print(f"📊 P-value: {p_value:.10f}")
print(f"\n🎯 Decision: {'REJECT H0' if p_value < 0.05 else 'FAIL TO REJECT H0'} (α = 0.05)")
print(f"💡 Conclusion: Traffic {'SIGNIFICANTLY' if p_value < 0.05 else 'DOES NOT significantly'} affects delivery time\n")

# Show mean delivery time by traffic
traffic_means = df.groupby('Traffic')['Delivery_Time'].mean().sort_values()
print("Average Delivery Time by Traffic:")
for traffic, time in traffic_means.items():
    print(f"  {traffic:15s}: {time:.2f} minutes")
print("=" * 70)
```

<h3 style="color:#007bff;">Chi-Square Test: Feature Independence</h3>

```python
from scipy.stats import chi2_contingency

# Test independence of Weather and Traffic
contingency_table = pd.crosstab(df['Weather'], df['Traffic'])
chi2, p_value, dof, expected = chi2_contingency(contingency_table)

print("=" * 70)
print("HYPOTHESIS TEST 3: Are Weather and Traffic Independent?")
print("=" * 70)
print(f"\n📝 Null Hypothesis (H0): Weather and Traffic are independent")
print(f"📝 Alternative Hypothesis (H1): Weather and Traffic are dependent\n")
print(f"🔬 Test: Chi-Square Test of Independence")
print(f"📊 Chi-Square statistic: {chi2:.4f}")
print(f"📊 Degrees of freedom: {dof}")
print(f"📊 P-value: {p_value:.10f}")
print(f"\n🎯 Decision: {'REJECT H0' if p_value < 0.05 else 'FAIL TO REJECT H0'} (α = 0.05)")
print(f"💡 Conclusion: Weather and Traffic are {'DEPENDENT' if p_value < 0.05 else 'INDEPENDENT'}")
print(f"💡 Implication: {'Model should consider their interaction' if p_value < 0.05 else 'Can treat as separate features'}\n")
print("=" * 70)
```

<h3 style="color:#007bff;">Normality Test: Delivery Time Distribution</h3>

```python
from scipy.stats import shapiro, normaltest

# Shapiro-Wilk test (sample first 5000 due to large dataset)
sample = df['Delivery_Time'].sample(5000, random_state=42)
stat, p_value = shapiro(sample)

print("=" * 70)
print("HYPOTHESIS TEST 4: Is Delivery Time Normally Distributed?")
print("=" * 70)
print(f"\n📝 Null Hypothesis (H0): Delivery time follows a normal distribution")
print(f"📝 Alternative Hypothesis (H1): Delivery time does not follow normal distribution\n")
print(f"🔬 Test: Shapiro-Wilk Test (sample of 5000)")
print(f"📊 Test statistic: {stat:.6f}")
print(f"📊 P-value: {p_value:.10f}")
print(f"\n🎯 Decision: {'REJECT H0' if p_value < 0.05 else 'FAIL TO REJECT H0'} (α = 0.05)")
print(f"💡 Conclusion: Delivery time {'IS NOT' if p_value < 0.05 else 'IS'} normally distributed")
print(f"💡 Implication: {'Consider log transformation or use non-parametric methods' if p_value < 0.05 else 'Parametric methods are appropriate'}\n")

# Visualize
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Histogram with normal curve overlay
from scipy.stats import norm
axes[0].hist(df['Delivery_Time'], bins=50, density=True, alpha=0.7, color='skyblue', edgecolor='black')
mu, std = df['Delivery_Time'].mean(), df['Delivery_Time'].std()
x = np.linspace(df['Delivery_Time'].min(), df['Delivery_Time'].max(), 100)
axes[0].plot(x, norm.pdf(x, mu, std), 'r-', linewidth=2, label=f'Normal(μ={mu:.1f}, σ={std:.1f})')
axes[0].set_title('Delivery Time Distribution vs Normal Distribution', fontsize=12, fontweight='bold')
axes[0].set_xlabel('Delivery Time (minutes)')
axes[0].set_ylabel('Density')
axes[0].legend()
axes[0].grid(True, alpha=0.3)

# Q-Q plot
from scipy.stats import probplot
probplot(df['Delivery_Time'], dist="norm", plot=axes[1])
axes[1].set_title('Q-Q Plot: Delivery Time vs Normal Distribution', fontsize=12, fontweight='bold')
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
print("=" * 70)
```

### 5.2. Cost & Efficiency Analysis 💰

<h3 style="color:#007bff;">Operational Cost Estimation</h3>

```python
# Cost assumptions (per delivery)
COST_PER_MINUTE = 0.50  # $0.50 per minute (agent wages, fuel, etc.)
COST_PER_KM = 0.30      # $0.30 per km (fuel, maintenance)
PENALTY_COST_SLA_MISS = 5.00  # $5 penalty for missing SLA

# Calculate costs
df['Time_Cost'] = df['Delivery_Time'] * COST_PER_MINUTE
df['Distance_Cost'] = df['Distance_km'] * COST_PER_KM
df['SLA_Penalty'] = df['SLA_Performance'].apply(lambda x: PENALTY_COST_SLA_MISS if x == 'Missed (>150 min)' else 0)
df['Total_Cost'] = df['Time_Cost'] + df['Distance_Cost'] + df['SLA_Penalty']

# Calculate efficiency metrics
df['Cost_Per_Km'] = df['Total_Cost'] / df['Distance_km']
df['Time_Per_Km'] = df['Delivery_Time'] / df['Distance_km']

# Summary statistics
print("=" * 70)
print("OPERATIONAL COST ANALYSIS")
print("=" * 70)
print(f"\n💰 Cost Breakdown (Average per Delivery):")
print(f"  Time Cost: ${df['Time_Cost'].mean():.2f}")
print(f"  Distance Cost: ${df['Distance_Cost'].mean():.2f}")
print(f"  SLA Penalty: ${df['SLA_Penalty'].mean():.2f}")
print(f"  Total Cost: ${df['Total_Cost'].mean():.2f}")
print(f"\n📊 Total Operational Metrics:")
print(f"  Total Deliveries: {len(df):,}")
print(f"  Total Cost: ${df['Total_Cost'].sum():,.2f}")
print(f"  Total Penalties: ${df['SLA_Penalty'].sum():,.2f}")
print(f"  Penalty Rate: {(df['SLA_Penalty'] > 0).sum()/len(df)*100:.1f}%")

# Visualize
fig, axes = plt.subplots(2, 2, figsize=(16, 10))

# Cost distribution
axes[0, 0].hist(df['Total_Cost'], bins=50, color='skyblue', edgecolor='black', alpha=0.7)
axes[0, 0].axvline(df['Total_Cost'].mean(), color='red', linestyle='--', linewidth=2, label=f'Mean: ${df["Total_Cost"].mean():.2f}')
axes[0, 0].set_title('Distribution of Total Cost per Delivery', fontsize=12, fontweight='bold')
axes[0, 0].set_xlabel('Total Cost ($)')
axes[0, 0].set_ylabel('Frequency')
axes[0, 0].legend()
axes[0, 0].grid(axis='y', alpha=0.3)

# Cost by traffic
traffic_cost = df.groupby('Traffic')['Total_Cost'].mean().sort_values()
bars = axes[0, 1].bar(range(len(traffic_cost)), traffic_cost.values, 
                     color=['green', 'yellow', 'orange', 'red'][:len(traffic_cost)], alpha=0.7)
axes[0, 1].set_title('Average Cost by Traffic Condition', fontsize=12, fontweight='bold')
axes[0, 1].set_ylabel('Average Cost ($)')
axes[0, 1].set_xticks(range(len(traffic_cost)))
axes[0, 1].set_xticklabels(traffic_cost.index)
axes[0, 1].grid(axis='y', alpha=0.3)

for i, v in enumerate(traffic_cost.values):
    axes[0, 1].text(i, v + 0.5, f'${v:.2f}', ha='center', fontweight='bold')

# Cost by vehicle
vehicle_cost = df.groupby('Vehicle')['Total_Cost'].mean().sort_values()
axes[1, 0].bar(range(len(vehicle_cost)), vehicle_cost.values, color='teal', alpha=0.7)
axes[1, 0].set_title('Average Cost by Vehicle Type', fontsize=12, fontweight='bold')
axes[1, 0].set_ylabel('Average Cost ($)')
axes[1, 0].set_xticks(range(len(vehicle_cost)))
axes[1, 0].set_xticklabels(vehicle_cost.index)
axes[1, 0].grid(axis='y', alpha=0.3)

for i, v in enumerate(vehicle_cost.values):
    axes[1, 0].text(i, v + 0.5, f'${v:.2f}', ha='center', fontweight='bold')

# Efficiency: Cost per km
axes[1, 1].scatter(df['Distance_km'], df['Cost_Per_Km'], alpha=0.3, s=10, c=df['Traffic'].map({'Low': 0, 'Medium': 1, 'High': 2, 'Jam': 3}), cmap='RdYlGn_r')
axes[1, 1].set_title('Cost Efficiency: Cost per Km vs Distance', fontsize=12, fontweight='bold')
axes[1, 1].set_xlabel('Distance (km)')
axes[1, 1].set_ylabel('Cost per Km ($)')
axes[1, 1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print("=" * 70)
```

### 5.3. Optimization Opportunities 🎯

<h3 style="color:#007bff;">High-Impact Improvement Areas</h3>

```python
# Identify problem segments
df['Problem_Segment'] = 'Normal'
df.loc[(df['Traffic'] == 'Jam') & (df['Distance_km'] > 15) & (df['Weather'].isin(['Fog', 'Cloudy'])), 'Problem_Segment'] = 'High Risk'
df.loc[(df['Traffic'].isin(['High', 'Jam'])) & (df['Distance_km'] > 10), 'Problem_Segment'] = 'Medium Risk'

# Calculate impact
segment_analysis = df.groupby('Problem_Segment').agg({
    'Delivery_Time': ['mean', 'count'],
    'Total_Cost': 'mean',
    'SLA_Penalty': 'sum'
}).round(2)
segment_analysis.columns = ['Avg_Delivery_Time', 'Count', 'Avg_Cost', 'Total_Penalties']

# Potential savings
baseline_time = df['Delivery_Time'].mean()
high_risk = df[df['Problem_Segment'] == 'High Risk']
medium_risk = df[df['Problem_Segment'] == 'Medium Risk']

if len(high_risk) > 0:
    high_risk_excess = (high_risk['Delivery_Time'].mean() - baseline_time) * len(high_risk)
    high_risk_savings = high_risk_excess * COST_PER_MINUTE
else:
    high_risk_excess = 0
    high_risk_savings = 0

if len(medium_risk) > 0:
    medium_risk_excess = (medium_risk['Delivery_Time'].mean() - baseline_time) * len(medium_risk)
    medium_risk_savings = medium_risk_excess * COST_PER_MINUTE * 0.5  # Assume 50% improvement
else:
    medium_risk_excess = 0
    medium_risk_savings = 0

fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Segment distribution
segment_counts = df['Problem_Segment'].value_counts()
colors_seg = {'Normal': '#28a745', 'Medium Risk': '#ffc107', 'High Risk': '#dc3545'}
axes[0].pie(segment_counts.values, labels=segment_counts.index, autopct='%1.1f%%',
           colors=[colors_seg.get(x, 'gray') for x in segment_counts.index], startangle=90,
           textprops={'fontsize': 11, 'fontweight': 'bold'})
axes[0].set_title('Delivery Segmentation by Risk Level', fontsize=14, fontweight='bold')

# Average delivery time by segment
segment_times = df.groupby('Problem_Segment')['Delivery_Time'].mean().sort_values()
bars = axes[1].barh(range(len(segment_times)), segment_times.values,
                   color=[colors_seg.get(x, 'gray') for x in segment_times.index], alpha=0.7)
axes[1].set_title('Average Delivery Time by Risk Segment', fontsize=14, fontweight='bold')
axes[1].set_xlabel('Average Delivery Time (minutes)')
axes[1].set_yticks(range(len(segment_times)))
axes[1].set_yticklabels(segment_times.index)
axes[1].grid(axis='x', alpha=0.3)

for i, v in enumerate(segment_times.values):
    axes[1].text(v + 2, i, f'{v:.1f} min', va='center', fontweight='bold')

plt.tight_layout()
plt.show()

print("=" * 70)
print("OPTIMIZATION OPPORTUNITIES")
print("=" * 70)
print(f"\n📊 Segment Analysis:")
print(segment_analysis.to_string())
print(f"\n💡 Improvement Potential:")
print(f"\n1️⃣ High Risk Segment ({len(high_risk):,} deliveries):")
print(f"   Current avg time: {high_risk['Delivery_Time'].mean():.1f} minutes" if len(high_risk) > 0 else "   No high risk deliveries")
print(f"   Excess time: {high_risk_excess:.0f} minutes" if len(high_risk) > 0 else "")
print(f"   Potential savings: ${high_risk_savings:,.2f}" if len(high_risk) > 0 else "")
print(f"\n2️⃣ Medium Risk Segment ({len(medium_risk):,} deliveries):")
print(f"   Current avg time: {medium_risk['Delivery_Time'].mean():.1f} minutes" if len(medium_risk) > 0 else "   No medium risk deliveries")
print(f"   Potential savings (50% improvement): ${medium_risk_savings:,.2f}" if len(medium_risk) > 0 else "")
print(f"\n🎯 Total Optimization Potential: ${high_risk_savings + medium_risk_savings:,.2f}")
print("=" * 70)
```
